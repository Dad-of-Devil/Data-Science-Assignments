######Numerical Columns####

```python
num_cols = df.select_dtypes(include=['int', 'float'])
```

```python
print(num_cols.columns)
```

#####Basics Statistics#####

```python
num_cols.mean()
```

###Mean (Average)####
The mean represents the average value of the dataset. It gives a general idea of the overall level of the data. However, the mean can be influenced by extreme values (outliers), so it may not always reflect the true center of the data.

```python
 num_cols.median()
```

####Median#####
The median is the middle value when the data is arranged in order. It is a better measure of central tendency when the dataset contains outliers, as it is not affected by extremely high or low values.

```python
 num_cols.mode().iloc[0]
```

###Mode####
The mode is the most frequently occurring value in the dataset. It helps identify the most common value or category and is especially useful for understanding patterns in categorical or repetitive data.

```python
num_cols.std()
```

#####Standard Deviation#####
Standard deviation measures how spread out the data values are from the mean. A low standard deviation indicates that the data points are close to the mean, while a high standard deviation shows that the data is widely spread out.

#####Histogram####

```python
import matplotlib.pyplot as plt

num_cols.hist(figsize=(8,8))
plt.show()
```

####Histogram####

-----Shows distribution of numerical data
-----Helps identify skewness (left/right)
-----Reveals data concentration (peaks)
-----Helps detect outliers or gaps

###Box Plot####

```python
import seaborn as sns

plt.figure(figsize=(12,8))
sns.boxplot(data=num_cols)
plt.show()
```

It Displays spread of data (IQR) , here IQR  is in the middle 50% data .
Shows median (center line)
Identifies outliers (points outside whiskers). Here the  point which are  away from the box these are the outliers.
Indicates data variability (wide/narrow box)

####Categorical columns ####

```python
categ_cols = df.select_dtypes(include=['object'])
print(categ_cols.columns)
```

###Box PLot ###

```python
for col in categ_cols.columns:
    df[col].value_counts().plot(kind='bar')
    plt.title(col)
    plt.show()
```

It shows the category which is more and which is less . And the what is the current trend based on the data.

--Shows frequency of categories.
--Helps compare different categories.
--Identifies most and least frequent values.
--Useful for spotting trends or imbalance.
