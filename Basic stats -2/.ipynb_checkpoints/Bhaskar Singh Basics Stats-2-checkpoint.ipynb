```python
import pandas as pd

df1 = pd.read_csv("Patient_Data.csv")
df2 = pd.read_csv("Billing_Data.csv")
```

```python
df1
```

```python
df1.info()
df2.head()
```

```python
df1 = df1[['PatientID', 'Department', 'Doctor', 'BillAmount']]
```

```python
df1
```

```python
df1 = df1.drop(columns=['ReceptionistID', 'CheckInTime'], errors='ignore')
```

```python
df1
```

```python
dept_bill = df1.groupby('Department')['BillAmount'].sum()
print(dept_bill)
```

```python
df1 = df1.drop_duplicates(subset='PatientID')
```

```python
df1
```

```python
####Handling Missing Values####
mean_bill = df1['BillAmount'].mean()
df1['BillAmount'].fillna(mean_bill, inplace=True)
```

```python
df1
```

####Merging Datasets####

```python
new_df = pd.merge(df1, df2, on='PatientID', how='inner')
```

```python
new_df
```

```python
new_patients = pd.DataFrame({
    'PatientID': [101, 102],
    'Department': ['Cardiology', 'Neurology'],
    'Doctor': ['Dr A', 'Dr B'],
    'BillAmount': [5000, 7000]
})

new_df = pd.concat([new_df, new_patients], axis=0)
```

```python
new_df
```

```python
####Adding New Billing Columns ####
```

```python
new_columns = pd.DataFrame({
    'InsuranceCovered': [True] * len(new_df),
    'FinalAmount': new_df['BillAmount'] * 0.9
})

new_df = pd.concat([new_df, new_columns], axis=1)
```
